# Legacy visualization example

```python
RANDOM_SEED = 42
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from lilypond import Basin
basin = Basin.from_data(X_scaled, random_seed=RANDOM_SEED, verbose=False)
```

```python
basin.legacy_pond() \
    .visualize_distance_map();
```

```python
basin.legacy_pond() \
    .visualize_activation_map();
```
