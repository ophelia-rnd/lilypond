# Color pads by a selected feature

```python
RANDOM_SEED = 42
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from lilypond import Basin
basin = Basin.from_data(X_scaled, random_seed=RANDOM_SEED, verbose=False).prepare()
```

```python
basin.pond() \
    .pad_layer(gap="nogap") \
    .petal_layer() \
    .visualize();
```

```python
basin.pond() \
    .pad_layer(gap="nogap", colorscale_source_feature_idx=0) \
    .visualize();
```

```python
basin.pond() \
    .pad_layer(gap="nogap", colorscale_source_feature_idx=1) \
    .visualize();
```

```python
basin.pond() \
    .pad_layer(gap="nogap", colorscale_source_feature_idx=1, colorscale="plasma") \
    .visualize();
```
