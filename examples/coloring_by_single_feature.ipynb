# Color pads by a selected feature

```python
RANDOM_SEED = 42
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from lilypond import Basin
basin = Basin.from_data_online(X_scaled, random_seed=RANDOM_SEED, verbose=False)
```

```python
basin.pond() \
    .pad_layer() \
    .petal_layer() \
    .visualize();
```

```python
basin.pond() \
    .pad_layer(gap="nogap", color_by="distance", colorscale="plasma") \
    .visualize(title="Color by distance");
```

```python
basin.pond() \
    .pad_layer(gap="nogap", color_by=dict(feature_idx=0), colorscale="plasma") \
    .visualize(title="Color by Feature 1");
```

```python
basin.pond() \
    .pad_layer(gap="nogap", color_by=dict(feature_idx=1), colorscale="plasma") \
    .visualize(title="Color by Feature 2");
```
