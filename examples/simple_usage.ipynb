# Simple `lilypond` usage

```
RANDOM_SEED = 42
```

```
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```
from lilypond import Basin
basin = Basin.from_data(X_scaled, random_seed=RANDOM_SEED, verbose=False).prepare()
```

```
basin.pond() \
    .pad_layer() \
    .petal_layer() \
    .visualize();
```

```
basin.pond(style_name="iceflock") \
    .pad_layer() \
    .petal_layer() \
    .visualize();
```

```
basin.pond() \
    .pad_layer() \
    .attraction_layer(X_scaled) \
    .visualize();
```

```
import numpy as np
X_scaled_QE = np.linalg.norm(basin.som.quantization(X_scaled) - X_scaled, axis=1)
```

```
basin.pond() \
    .pad_layer() \
    .attraction_layer(X_scaled, jitter_amount=.3, marker=dict(opacity=.75, size=10, color=X_scaled_QE, colorscale="Spectral_r", line=dict(width=1, color="black"))) \
    .visualize();
```

```
basin.pond(style_name="iceflock") \
    .pad_layer() \
    .attraction_layer(X_scaled, jitter_amount=.3, marker=dict(opacity=.85, size=10, color=X_scaled_QE, colorscale="Spectral_r", line=dict(width=1, color="black"))) \
    .visualize();
```
