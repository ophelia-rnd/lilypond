# Simple usage of `lilypond`

```python
RANDOM_SEED = 42
VERBOSE = False
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from lilypond import Basin
basin = Basin.from_data_online(X_scaled, random_seed=RANDOM_SEED, verbose=VERBOSE)
```

```python
basin.pond() \
    .pad_layer() \
    .petal_layer() \
    .visualize();
```

```python
basin.pond(base_style="iceflock") \
    .pad_layer() \
    .petal_layer() \
    .visualize();
```

```python
basin.pond() \
    .pad_layer() \
    .attraction_layer(X_scaled) \
    .visualize();
```

```python
import numpy as np
X_scaled_QE = np.linalg.norm(basin.som_representation.som.quantization(X_scaled) - X_scaled, axis=1)
```

```python
attraction_marker = dict(opacity=.75, size=10, color=X_scaled_QE, colorscale="Spectral_r", line=dict(width=1, color="black"))
basin.pond() \
    .pad_layer() \
    .attraction_layer(X_scaled, jitter_amount=.25, marker=attraction_marker) \
    .visualize();
```

```python
basin.pond(base_style="iceflock") \
    .pad_layer() \
    .attraction_layer(X_scaled, jitter_amount=.25, marker=dict(attraction_marker, opacity=.85)) \
    .visualize();
```
