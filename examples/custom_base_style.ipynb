# Setting a custom base style

```python
RANDOM_SEED = 42
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from lilypond import Basin
basin = Basin.from_data(X_scaled, random_seed=RANDOM_SEED, verbose=False).prepare()
```

```python
from lilypond.pond_base_style import PondBaseStyle

modern_style = PondBaseStyle(
    water_colorscale="Teal",
    rhizome_colorscale="dense",
    pad_colorscale="Plasma",
    petal_colorscale="Greys",
    petal_marker_line=dict(width=0),
    petal_marker=dict(symbol="star", opacity=1),
    petal_halo_marker=None
)

basin.pond(base_style=modern_style) \
    .rhizome_layer() \
    .pad_layer() \
    .petal_layer() \
    .visualize();
```

```python
basin.pond(base_style=modern_style) \
    .rhizome_layer(colorscale="Greys", min_width=2, max_width=2) \
    .pad_layer(colorscale="Plasma_r") \
    .petal_layer(marker=dict(size=16, color="black")) \
    .visualize();
```
