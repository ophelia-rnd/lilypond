# Observing dual best-matching unit connections aka. the rhizome layer

```python
RANDOM_SEED = 42
VERBOSE = False
```

```python
# load dataset
from sklearn.datasets import load_iris
X, y = load_iris(return_X_y=True, as_frame=False)
# scale data
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
from lilypond import Basin
basin = Basin.from_data_online(X_scaled, random_seed=RANDOM_SEED, verbose=VERBOSE)
```

```python
basin.pond() \
    .rhizome_layer() \
    .pad_layer() \
    .visualize();
```

```python
basin.pond() \
    .rhizome_layer(violations_only=True) \
    .pad_layer() \
    .visualize();
```

```python
basin.pond() \
    .rhizome_layer(violations_only=True, neighborhood="von-neumann") \
    .pad_layer() \
    .visualize();
```

```python
basin.pond() \
    .pad_layer() \
    .rhizome_layer() \
    .visualize();
```

```python
basin.pond() \
    .rhizome_layer(min_width=3, max_width=18, colorscale="jet") \
    .pad_layer() \
    .visualize();
```

```python
basin.pond() \
    .rhizome_layer(min_width=6, max_width=6, colorscale="jet") \
    .pad_layer() \
    .visualize();
```

```python
# plot the connections implied by an arbitrary data collection
import numpy as np
rng = np.random.default_rng(RANDOM_SEED)
sampled_indices = rng.choice(X_scaled.shape[0], size=25, replace=False)

basin.pond() \
    .pad_layer() \
    .rhizome_layer(X=X_scaled[sampled_indices]) \
    .visualize();
```
